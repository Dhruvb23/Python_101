```python
sales_data = ["1200", "1500", "1100", "1800", "1600", "1400", "1700"]
sales_num=[]
for i in sales_data:
    if i not in sales_num:
        sales_num.append(int(i))
print(sales_num)
print(sales_data)
for i,sale in enumerate(sales_num):
    print("Day",i+1,":","rs.", sale)

total= sum(sales_num)
avg=sum(sales_num)/len(sales_num)

print("sum of the sales: rs.",total)
print("avg of the sales: rs.",round(avg,2))
```
